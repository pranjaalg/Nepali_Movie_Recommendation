```python
import numpy as np
import pandas as pd
import seaborn as sns
import nltk
import re
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
```

```python
nltk.download('stopwords')
nltk.download('wordnet')
```

```python
df=pd.read_csv('/content/top_100_nepali_movies.csv')
```

```python
df.isnull().sum()
```

```python
stop_words=set(nltk.corpus.stopwords.words('nepali'))
leammatizer=WordNetLemmatizer()
```

```python
df
```

```python
df['combined']=(df['Title'])*2 +' '+ df['Description']+' '+df['Tags']
```

```python
df
```

```python
def preprocessing_nepali(text):
  cleaned_text = re.sub(r'[^\u0900-\u097F\s]', '', text)

  cleaned_text = re.sub(r'\s+', ' ', cleaned_text).strip()
  words=cleaned_text.split()
  words=[word for word in words if word not in stop_words]
  words=[leammatizer.lemmatize(word) for word in words]
  return ' '.join(words)


```

```python
df['combined']=df['combined'].apply( preprocessing_nepali)
```

```python
indices = pd.Series(df.index, index=df['Title']).drop_duplicates()
```

```python
indices
```

VECTORIZATION


```python
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
tfidf=TfidfVectorizer(max_features=300,ngram_range=(1,2))
```

```python
tfidf_matrix=tfidf.fit_transform(df['combined'])
```

```python
tfidf_matrix
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
def recommend(Title,n=5):
  if Title not in indices:
    return ['Movie not Found']
  idx=indices[Title]
  sim_score=cosine_similarity(tfidf_matrix[idx],tfidf_matrix).flatten()
  sim_id=sim_score.argsort()[::-1][1:n+1]
  return df['Title'].iloc[sim_id]
```

```python
recommend('छक्का पञ्जा')
```
